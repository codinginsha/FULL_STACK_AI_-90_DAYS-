```python
#Assignment 4 : Create a new list containing the squares of the first 10 positive integers using a list comprehension. Print the new list.
import builtins
list = []
for i in builtins.range(1, 11):
    list.append(i**2)
    
print("Squares of the first 10 positive integers:", list)
```

```python
#Assignment 5 : Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.
improt random

# Create a list of random numbers
random_numbers = [random.randint(1, 10) for _ in range(20)]

# Sort the list in ascending order
random_numbers.sort()

# Print the sorted list
print("List in ascending order:", random_numbers)

# Sort the list in descending order
random_numbers.sort(reverse=True)

# Print the sorted list
print("List in descending order:", random_numbers)

# Remove duplicates from the list
random_numbers = list(set(random_numbers))

# Print the modified list
print("List with duplicates removed:", random_numbers)
```

```python
#Assignment 7 : Create a list of dictionaries where each dictionary represents a student with keys 'name' and 'score'. Sort the list of dictionaries by the 'score' in descending order and print the sorted list.
students = [
    {'name': 'Alice', 'score': 85},
    {'name': 'Bob', 'score': 90},
    {'name': 'Charlie', 'score': 78},
    {'name': 'David', 'score': 95}
]
 
# Sort the list of dictionaries by 'score' in descending order
students.sort(reverse=True, key=lambda x: x['score'])
print(students)
```
